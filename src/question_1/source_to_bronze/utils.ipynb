```python
import re

from pyspark.sql.functions import current_date, udf
from pyspark.sql.types import StringType
```

```python
def read_csv(path):
    return spark.read.option("header", True).csv(path)
```

```python
def write_csv(df, path):
    df.write.format("csv") \
        .option("header", True) \
        .mode("overwrite") \
        .save(path)
```

```python
def camel_to_snake(name):
    name = re.sub(r"([A-Z]+)([A-Z][a-z])", r"\1_\2", name)
    name = re.sub(r"([a-z0-9])([A-Z])", r"\1_\2", name)
    return name.lower()


camel_to_snake_udf = udf(camel_to_snake, StringType())


def convert_columns(df):
    new_columns = []

    for name in df.columns:
        new_columns.append(camel_to_snake(name))

    for old_name, new_name in zip(df.columns, new_columns):
        df = df.withColumnRenamed(old_name, new_name)

    return df
```

```python
def add_load_date(df):
    return df.withColumn("load_date", current_date())
```
